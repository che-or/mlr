```python
import pandas as pd
import numpy as np
from load_gamelogs import _load_gamelogs, _load_player_types
from download_gamelogs import _download_gamelogs
```

```python
_download_gamelogs()
```

```python
league = 'mlr'
season = 12

df = _load_gamelogs(season, league, False)
players = _load_player_types(season, league, False)
```

```python
hitters = df['Hitter ID'].unique()
pitchers = df[df['Pitcher ID'].notna()]['Pitcher ID'].unique()

all_ids = np.unique(np.concatenate((hitters, pitchers)))
```

```python
all_ids = pd.DataFrame(all_ids, columns = ['ID'])
```

```python
all_ids['Hitter?'] = all_ids['ID'].isin(hitters)
all_ids['Pitcher?'] = all_ids['ID'].isin(pitchers)
```

```python
season_players = all_ids.merge(players, on = 'ID', how = 'left')
```

```python
season_players[['Pitching Type', 'Pitching Bonus']] = season_players['Pitching Type'].str.split('-', expand = True)
season_players = season_players[['ID', 'Name', 'Batting Type', 'Pitching Type', 'Pitching Bonus', 'Handedness', 'Position', 'Hitter?', 'Pitcher?']]
season_players = season_players.rename(columns = {'ID': 'Player ID', 'Position': 'Primary Position'})
```

```python
season_players.to_csv('players.csv', index = False, encoding = 'utf-8-sig')
```
